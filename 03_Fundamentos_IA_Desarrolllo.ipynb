# 1. ¿Qué es la Lógica?

La lógica es la disciplina que estudia las reglas del pensamiento correcto. Permite:

- identificar relaciones de causa y efecto  
- detectar contradicciones  
- deducir conclusiones válidas  
- estructurar procesos de decisión  
- resolver problemas mediante pasos claros  

En programación, la lógica es indispensable porque un computador solo entiende:

- Verdadero  
- Falso  
- Instrucciones secuenciales  
- Condiciones  
- Reglas precisas

# 2. Tipos de Razonamiento

## 2.1 Razonamiento Deductivo (general → particular)

**Definición:**  
Parte de lo general para llegar a algo particular.
Si las premisas son verdaderas, la conclusión siempre es verdadera.

**Ejemplo:**  
- Premisa: Todos los mamíferos tienen pulmones.  
- Premisa: El perro es mamífero.  
- Conclusión: El perro tiene pulmones.  

**Analogía:**  
Si una receta dice: “Si metes agua al congelador, se convierte en hielo”, entonces esperarás hielo al meter agua.

---

## 2.2 Razonamiento Inductivo (particular → general)

**Definición:**  
Parte de casos particulares para proponer una regla general.
La conclusión es probable, pero no siempre 100% segura.

**Ejemplo:**  
- Ayer el sol salió por el este  
- Hoy salió por el este  
→ Probablemente siempre sale por el este  

**Analogía:**  
Si pruebas 3 dulces de una bolsa y todos son de fresa, concluyes que probablemente todos lo son.

# 8. Operadores Lógicos

Los **operadores lógicos** permiten conectar proposiciones (enunciados que pueden ser verdaderos o falsos) para crear expresiones más complejas y evaluar su valor de verdad.  
Son fundamentales en lógica matemática, programación, bases de datos y razonamiento formal.

---

## 1. AND (Conjunción lógica)

**Definición:**  
El resultado es **verdadero solo si ambas proposiciones son verdaderas**.

**Símbolos:**  
- AND  
- ∧  
- && (programación)

**Tabla de verdad:**

| A | B | A AND B |
|---|---|----------|
| V | V | V        |
| V | F | F        |
| F | V | F        |
| F | F | F        |

**Ejemplo:**  
“Tengo dinero **y** la tienda está abierta.”

---

## 2. OR (Disyunción lógica)

**Definición:**  
El resultado es **verdadero si al menos una proposición es verdadera**.

**Símbolos:**  
- OR  
- ∨  
- || (programación)

**Tabla de verdad:**

| A | B | A OR B |
|---|---|---------|
| V | V | V       |
| V | F | V       |
| F | V | V       |
| F | F | F       |

**Ejemplo:**  
“Puedo ir en bus **o** en taxi.”

---

## 3. NOT (Negación lógica)

**Definición:**  
Invierte el valor lógico de una proposición:  
- Si A es verdadera → NOT A es falsa.  
- Si A es falsa → NOT A es verdadera.

**Símbolos:**  
- NOT  
- ¬  
- ! (programación)

**Tabla de verdad:**

| A | NOT A |
|---|--------|
| V | F      |
| F | V      |

**Ejemplo:**  
“No es de día.”

---

## Resumen rápido

| Operador | Nombre | Cuándo es verdadero |
|----------|--------|---------------------|
| AND (∧) | Conjunción | Solo si A y B son verdaderas |
| OR (∨) | Disyunción | Si al menos una es verdadera |
| NOT (¬) | Negación | Si invierte el valor |

# Ejercicios

# 9.1 Número PAR

## Lenguaje natural
El programa debe recibir un número entero `n` y determinar si es **par**.  
Un número es par cuando, al dividirlo entre 2, el residuo es igual a 0.

## Pseudocódigo
```
INICIO
    ESCRIBIR "Ingrese un número entero:"
    LEER n

    SI n % 2 = 0 ENTONCES
        ESCRIBIR "El número es PAR"
    SINO
        ESCRIBIR "El número es IMPAR"
    FIN SI
FIN
```

## Diagrama textual
```
Inicio
↓
Mostrar "Ingrese un número entero"
↓
Leer n
↓
¿n % 2 = 0?
    ├─ Sí → Mostrar "El número es PAR"
    └─ No → Mostrar "El número es IMPAR"
↓
Fin
```

---

# 9.2 Número IMPAR

## Lenguaje natural
Un número es impar cuando el residuo de dividirlo entre 2 **no es 0**.

## Pseudocódigo
```
INICIO
    ESCRIBIR "Ingrese un número entero:"
    LEER n

    SI n % 2 <> 0 ENTONCES
        ESCRIBIR "El número es IMPAR"
    SINO
        ESCRIBIR "El número es PAR"
    FIN SI
FIN
```

## Diagrama textual
```
Inicio
↓
Leer n
↓
¿n % 2 <> 0?
    ├─ Sí → Mostrar "El número es IMPAR"
    └─ No → Mostrar "El número es PAR"
↓
Fin
```

# 9.6 División con validación

## Lenguaje natural
El programa divide `num / den`, pero antes verifica que `den` **no sea 0**.

## Pseudocódigo
```
INICIO
    ESCRIBIR "Ingrese el numerador:"
    LEER num

    ESCRIBIR "Ingrese el denominador:"
    LEER den

    SI den = 0 ENTONCES
        ESCRIBIR "ERROR: No se puede dividir entre cero."
    SINO
        resultado ← num / den
        ESCRIBIR "El resultado es: ", resultado
    FIN SI
FIN
```

## Diagrama textual
```
Inicio
↓
Leer num
↓
Leer den
↓
¿den = 0?
    ├─ Sí → Mostrar "ERROR"
    └─ No → resultado ← num / den
            Mostrar resultado
↓
Fin
```

# 9.7 Sacar plata de un cajero

## Lenguaje natural
Para que un retiro sea posible se deben cumplir todas las reglas:

1. `monto ≤ saldo`  
2. `monto ≤ fondosCajero`  
3. `monto % 10 = 0`

## Pseudocódigo
```
INICIO
    ESCRIBIR "Ingrese el saldo:"
    LEER saldo

    ESCRIBIR "Ingrese fondos del cajero:"
    LEER fondosCajero

    ESCRIBIR "Ingrese monto a retirar:"
    LEER monto

    SI (monto <= saldo) Y (monto <= fondosCajero) Y (monto % 10 = 0) ENTONCES
        saldo ← saldo - monto
        fondosCajero ← fondosCajero - monto
        ESCRIBIR "Retiro exitoso"
        ESCRIBIR "Nuevo saldo: ", saldo
    SINO
        ESCRIBIR "No es posible retirar"
    FIN SI
FIN
```

## Diagrama textual
```
Inicio
↓
Leer saldo
↓
Leer fondosCajero
↓
Leer monto
↓
¿monto <= saldo?
    ├─ No → Mostrar "No es posible retirar" → Fin
    └─ Sí →
        ¿monto <= fondosCajero?
            ├─ No → Mostrar "No es posible retirar" → Fin
            └─ Sí →
                ¿monto % 10 = 0?
                    ├─ No → Mostrar "No es posible retirar" → Fin
                    └─ Sí →
                        saldo ← saldo - monto
                        fondosCajero ← fondosCajero - monto
                        Mostrar "Retiro exitoso"
↓
Fin
```

# 9.9 Sistema de acceso

## Lenguaje natural
Acceso permitido solo si:

- clave = "1234"  
- usuarioActivo = TRUE  
- rol = admin o supervisor  

## Pseudocódigo
```
INICIO
    LEER clave
    LEER usuarioActivo
    LEER rol

    SI (clave = "1234") Y (usuarioActivo = TRUE) Y (rol = "admin" O rol = "supervisor") ENTONCES
        ESCRIBIR "Acceso concedido"
    SINO
        ESCRIBIR "Acceso denegado"
    FIN SI
FIN
```

## Diagrama textual
```
Inicio
↓
Leer clave
↓
Leer usuarioActivo
↓
Leer rol
↓
¿clave = "1234"?
    ├─ No → Mostrar "Acceso denegado"
    └─ Sí →
        ¿usuarioActivo = TRUE?
            ├─ No → Mostrar "Acceso denegado"
            └─ Sí →
                ¿rol es admin o supervisor?
                    ├─ No → Mostrar "Acceso denegado"
                    └─ Sí → Mostrar "Acceso concedido"
↓
Fin
```

# 9.10 Semáforo

## Lenguaje natural
El programa debe indicar qué hacer según el color ingresado.

## Pseudocódigo
```
INICIO
    LEER color

    SI color = "rojo" ENTONCES
        ESCRIBIR "Detener"
    SINO
        SI color = "amarillo" ENTONCES
            ESCRIBIR "Precaución"
        SINO
            SI color = "verde" ENTONCES
                ESCRIBIR "Avanzar"
            SINO
                ESCRIBIR "Color no válido"
            FIN SI
        FIN SI
    FIN SI
FIN
```

## Diagrama textual
```
Inicio
↓
Leer color
↓
¿color = "rojo"?
    ├─ Sí → Mostrar "Detener"
    └─ No →
        ¿color = "amarillo"?
            ├─ Sí → Mostrar "Precaución"
            └─ No →
                ¿color = "verde"?
                    ├─ Sí → Mostrar "Avanzar"
                    └→ No → Mostrar "Color no válido"
↓
Fin
```
